```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("employees_100.csv")
print(df)
```

## Cleaning Data


```python
# Missing Values
print(df.isnull().sum())
```

```python
#Missing values ko remove karna
print(df.dropna(subset=['city',"age"]))
```

```python
df.head()
```

```python
# fill the value in column
df['age'] = df['age'].fillna(df['age'].mean())

print(df['age'])
```

```python
print(df.isnull().sum())
```

```python
# df['salary']=df['salary'].str.replace(",", "", regex=False) # ye (,) gatane ko
df['salary']=pd.to_numeric(df['salary'],errors='coerce')      # ye numaric banane ko
df['salary']=df['salary'].fillna(df['salary'].median())       # ye fill krne ko

df['salary'].dtype

```

```python
print(df.isnull().sum())
```

```python
df['city'].value_counts()
```

```python
df['city']=df['city'].fillna("Gwalior")
print(df.isnull().sum())
df['city'].value_counts()
```

```python
# Check for Duplicate
df.duplicated(subset=["employee_id"]).sum() 
# delete duplicate value
df.drop_duplicates(subset=["employee_id"])

df.head()
```

```python
# change normal time to real time
df["joining_date"]=pd.to_datetime(df["joining_date"])

df.head()
```

### Renaming column

```python
df.rename(columns={"name":"employee_name"}, inplace=True)
df.head()
```

## Modifying Data & Creating New Columns

```python
df['annual_salary']=df['salary']*12
df.head()
```

#### Apply() Function

```python
def categorize(num):
    if num>5:
        return "Experience"
    elif num<=5:
        return "Learner"

df['experience_status']=df['experience'].apply(categorize)
df.head()    
```

#### Map() Function

```python
department_code={"Engineering":"ENG","Data Science":"DS","Marketing":"MKT"}
df["department_code"]=df["department"].map(department_code)

df.head()
```

## Grouping & Aggregation

##### process-->
######    split(by group)---->apply(mean)---->Combine(into a table)

```python
df['city'].unique()
```

```python
# .sum(), .mean(), .median(), .min(), .max(), .count(), .size(), .std()

print(df.groupby("city")['age'].sum())
print(df.groupby("city")['age'].mean())
print(df.groupby("city")['age'].median())
print(df.groupby("city")['age'].min())
print(df.groupby("city")['salary'].max())
print(df.groupby("city")['age'].count())
print(df.groupby("city")['age'].size())
print(df.groupby("city")['age'].std())
```

```python
df.groupby(['city','department'])['salary'].mean()
```

```python
# multi_groups

df.groupby(["department", "city"])["salary"].mean()   # hierarchical grouping

df.groupby("department")[["salary", "performance_score"]].mean()
```

```python
# agg_function.py

df.groupby("department").agg({
    "salary": ["mean", "max"],
    "performance_score": "mean",
    "employee_id": "count"
})
```

```python
# Named aggregation

department_summary = df.groupby("department").agg(
    average_salary=("salary", "mean"),
    highest_salary=("salary", "max"),
    employee_count=("employee_id", "count")
)

print(department_summary)
```
