## Selecting & Filtering Data

```python
print(df["salary"])   # one column -> a Series
print(df[["name", "department", "salary"]])    # multiple columns -> a DataFrame
```

#### Interactive: loc vs iloc


```python
df = df.reset_index()
df = df.set_index("employee_id")
print(df.head())
```

```python
# .loc[] → label/name ke basis par data access
# .iloc[] → position/number ke basis par data access

# 1
print("1\n",df.loc[264])

# 2
print("2\n",df.loc[264, "salary"])

# 3
print("3\n",df.loc[[264, 107, 171]])

# 4
print("4\n",df.loc[264, ["name", "salary"]])

# 5
print("5\n",df.iloc[0])

# 6
print("6\n",df.iloc[0:5])

# 7
print("7\n",df.iloc[0, 1])

# 8
print("8\n",df.iloc[0:5, 0:3])

# 9
print("9\n",df.iloc[:, 1])

# 10
print("10\n",df.iloc[:, 0:3])
```

#### boolean_filter(Condition)


```python
a=df[(df["experience"] > 3) & (df["performance_score"] > 7)]   # AND
print(a)
```
